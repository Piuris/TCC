# 1. Análise exploratória de dados

TCC — Detecção de fraudes em transações financeiras (IFRS Campus Canoas).

Este notebook usa o pacote `fraude`; os mesmos resultados são gerados por `python -m fraude eda`.

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
import pandas as pd
from IPython.display import Image, Markdown, display
pd.set_option("display.max_columns", 40)
DATASET = "creditcard"
REPORTS = ROOT / "reports" / DATASET
```

```python
from fraude.data import load_dataset
from fraude.eda import run_eda

ds = load_dataset(DATASET)
ds.summary()
```

```python
ds.X.describe().T
```

## Desbalanceamento

Um classificador que aprova todas as transações acerta mais de 99,8% — e não detecta nenhuma fraude.

```python
stats = run_eda(ds, REPORTS / 'eda')
print('Acurácia do classificador trivial:', stats['trivial_accuracy_pct'], '%')
print('Duplicatas:', stats['duplicates'])
display(Image(REPORTS / 'eda' / 'class_balance.png'))
```

## Valor das transações

```python
display(pd.DataFrame(stats['amount_by_class']))
display(Image(REPORTS / 'eda' / 'amount_by_class.png'))
```
